```python
from pathlib import Path
import cv2, numpy as np, matplotlib.pyplot as plt
R = Path.cwd() if (Path.cwd()/'images').exists() else Path.cwd().parent
im=cv2.imread(str(R/'images/source/coins.png'),0)
g_x=cv2.Sobel(im,cv2.CV_32F,1,0,ksize=3)
g_y=cv2.Sobel(im,cv2.CV_32F,0,1,ksize=3)
# TODO: gradient magnitude, direction, thresholds, Laplace
fig,axs=plt.subplots(1,3,figsize=(12,4))
for ax,im0,title in zip(axs,[im,g_x,g_y],['Вход','Gx','Gy']):ax.imshow(im0,cmap='gray');ax.set_title(title);ax.axis('off')
plt.show()
```
